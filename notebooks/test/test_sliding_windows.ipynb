```python

import numpy as np
import torch
```

```python
from monai.inferers.inferer import SlidingWindowInferer
```

```python
roi_size=[128,128,128]
inferer = SlidingWindowInferer(roi_size=roi_size, overlap=0.9, sw_batch_size=1)
image = torch.randn(1, 3, 256,256,256)
out  = inferer(inputs=image, network=lambda x: x.sigmoid())
print(((image.sigmoid()-out)**2).mean())
```
