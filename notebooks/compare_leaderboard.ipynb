```python
import pandas as pd
```

```python
data_path = 'BraTS-leaderboard-12-10-24.csv'
```

```python
df = pd.read_csv(data_path)
```

```python
df
```

```python
df['submission_status'].unique()
```

```python
#Filter df

df = df[(df['status'] == 'ACCEPTED') & (df['submission_status'] == 'SCORED') & (df['cases_evaluated'] == 219)]
df
```

```python
df.columns
```

```python
subset_cols = ['ROW_ID', 'ROW_VERSION', 'evaluationid', 'id', 'createdOn', 'name', \
       'submitterid', 'status', 'submission_status', 'cases_evaluated', \
       'LesionWise_Dice_ET', 'LesionWise_Dice_TC', 'LesionWise_Dice_WT', \
       'LesionWise_Hausdorff95_ET', 'LesionWise_Hausdorff95_TC', 'LesionWise_Hausdorff95_WT']
```

```python
df = df[subset_cols]
df
```

```python
df['LesionWise_Dice_WT_Rank'] = df['LesionWise_Dice_WT'].rank(ascending=True,pct=True)
df['LesionWise_Dice_TC_Rank'] = df['LesionWise_Dice_TC'].rank(ascending=True,pct=True)
df['LesionWise_Dice_ET_Rank'] = df['LesionWise_Dice_ET'].rank(ascending=True,pct=True)
```

```python
df['LesionWise_Hausdorff95_WT_Rank'] = df['LesionWise_Hausdorff95_WT'].rank(ascending=False,pct=True)
df['LesionWise_Hausdorff95_TC_Rank'] = df['LesionWise_Hausdorff95_TC'].rank(ascending=False,pct=True)
df['LesionWise_Hausdorff95_ET_Rank'] = df['LesionWise_Hausdorff95_ET'].rank(ascending=False,pct=True)
```

```python
df[df['LesionWise_Dice_WT_Rank'] > 0.99]['name']
```

```python
import matplotlib.pyplot as plt
```

```python
score_cols = ['LesionWise_Dice_WT', 'LesionWise_Hausdorff95_WT', 'LesionWise_Dice_TC', 'LesionWise_Hausdorff95_TC', 'LesionWise_Dice_ET', 'LesionWise_Hausdorff95_ET']
score_cols = [f"{col}_Rank" for col in score_cols] 
df[df['id'] == 9750300][score_cols].plot(kind='bar', figsize=(9, 7), grid=True)
# Set labels and title
plt.ylabel('Percentile Rank')
plt.xticks([])
plt.title('Leaderboard Percentile Ranks for PatchDiffusion')
plt.legend(loc='center left', bbox_to_anchor=(1.0, 0.5))

# Show plot
plt.show()
```
